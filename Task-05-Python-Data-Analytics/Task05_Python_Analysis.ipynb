```python
# Task 05 - Python for Data Analytics
# Load Superstore dataset

import pandas as pd
import numpy as np

# Load the CSV
df = pd.read_csv("SampleSuperstore.csv", encoding="latin1")

# Show the first 5 rows
df.head()
```

```python
# Basic info about the dataset
print("Shape (rows, columns):", df.shape)
print()
print("Column names:")
print(df.columns.tolist())
print()
print("Data types:")
print(df.dtypes)
print()
print("First 5 rows:")
df.head()
```

```python
# Convert Order Date and Ship Date from text to datetime
# Format is DD-MM-YYYY
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')

# Confirm the conversion
print("Order Date dtype:", df['Order Date'].dtype)
print("Ship Date dtype:", df['Ship Date'].dtype)
print()
print("Date range:")
print("Earliest order:", df['Order Date'].min())
print("Latest order:  ", df['Order Date'].max())
```

```python
# Filter 1: High-value orders in West region (same as SQL Query 9)
high_value_west = df[(df['Region'] == 'West') & (df['Sales'] > 1000)]
print("High-value orders in West:", len(high_value_west))
print()
print(high_value_west[['Order ID', 'Customer Name', 'Region', 'Sales', 'Profit']].head(10))
```

```python
# Filter 2: Loss-making orders (Profit < 0)
losses = df[df['Profit'] < 0]
print("Total loss-making orders:", len(losses))
print("Total loss amount: $", round(losses['Profit'].sum(), 2))
print()
print(losses[['Order ID', 'Customer Name', 'Category', 'Sales', 'Discount', 'Profit']].head(10))
```

```python
# Group by Region, calculate total Sales and Profit
region_summary = df.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Order_Count=('Order ID', 'count')
).round(2)

region_summary = region_summary.sort_values('Total_Sales', ascending=False)
print(region_summary)
```

```python
# Group by Region AND Category
region_category = df.groupby(['Region', 'Category']).agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).round(2)

print(region_category)
```

```python
# Top 10 customers by total sales
top_customers = df.groupby('Customer Name')['Sales'].sum().round(2)
top_customers = top_customers.sort_values(ascending=False).head(10)

print("Top 10 Customers by Total Sales:")
print(top_customers)
```

```python
# Create two separate DataFrames (like two SQL tables)
# 1) Orders table: one row per order line
orders = df[['Order ID', 'Customer ID', 'Customer Name', 'Sales', 'Profit', 'Order Date']].copy()

# 2) Customers table: one row per unique customer with segment info
customers = df[['Customer ID', 'Segment', 'Region']].drop_duplicates().copy()

print("Orders table shape:", orders.shape)
print("Customers table shape:", customers.shape)
print()
print("Customers table sample:")
print(customers.head())
```

```python
# Merge orders and customers on Customer ID (equivalent to SQL JOIN)
merged = pd.merge(orders, customers, on='Customer ID', how='left')

print("Merged shape:", merged.shape)
print()
print("Merged columns:", merged.columns.tolist())
print()
print(merged.head(10))
```

```python
# Analyze profit by segment using the MERGED dataset
segment_analysis = merged.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Order_Count=('Order ID', 'count'),
    Avg_Order_Value=('Sales', 'mean')
).round(2)

segment_analysis = segment_analysis.sort_values('Total_Sales', ascending=False)
print("Segment Performance:")
print(segment_analysis)
```

```python
# Pivot table: Sales by Region (rows) x Category (columns)
pivot_sales = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Category',
    aggfunc='sum'
).round(2)

print("Sales by Region and Category:")
print(pivot_sales)
```

```python
# Profit by Sub-Category (worst to best)
subcat_profit = df.groupby('Sub-Category')['Profit'].sum().round(2)
subcat_profit = subcat_profit.sort_values(ascending=True)

print("Sub-Category Profit (worst to best):")
print(subcat_profit)
```

```python
# Monthly sales trend
df['YearMonth'] = df['Order Date'].dt.to_period('M')
monthly = df.groupby('YearMonth').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).round(2)

print("Monthly Sales (last 12 months shown):")
print(monthly.tail(12))
```

```python
# Apply CASE-like logic using pd.cut()
df['Discount_Band'] = pd.cut(
    df['Discount'],
    bins=[-0.01, 0, 0.2, 0.4, 0.6, 1.0],
    labels=['No Discount', '1-20%', '21-40%', '41-60%', '60%+']
)

discount_analysis = df.groupby('Discount_Band', observed=True).agg(
    Order_Count=('Order ID', 'count'),
    Avg_Profit=('Profit', 'mean'),
    Total_Profit=('Profit', 'sum')
).round(2)

print("Discount Impact on Profit:")
print(discount_analysis)
```

```python
# Rank customers within each region by total sales (like SQL ROW_NUMBER with PARTITION)
customer_region = df.groupby(['Region', 'Customer Name'])['Sales'].sum().reset_index()
customer_region['Rank'] = customer_region.groupby('Region')['Sales'].rank(
    method='first', ascending=False
).astype(int)

# Top customer in each region
top_per_region = customer_region[customer_region['Rank'] == 1].sort_values('Region')
print("Top Customer per Region:")
print(top_per_region.to_string(index=False))
```

```python
# Descriptive statistics for numeric columns
print("Descriptive Statistics:")
print(df[['Sales', 'Quantity', 'Discount', 'Profit']].describe().round(2))
```

```python
# Correlation between numeric columns
correlation = df[['Sales', 'Quantity', 'Discount', 'Profit']].corr().round(3)
print("Correlation Matrix:")
print(correlation)
```

```python
# Save analysis results to CSV files for reproducibility
region_summary.to_csv("output_region_summary.csv")
segment_analysis.to_csv("output_segment_analysis.csv")
subcat_profit.to_csv("output_subcategory_profit.csv")
discount_analysis.to_csv("output_discount_impact.csv")

print("Saved 4 CSV files:")
print("- output_region_summary.csv")
print("- output_segment_analysis.csv")
print("- output_subcategory_profit.csv")
print("- output_discount_impact.csv")
```

## Key Findings

### 1. Regional Performance
- West leads in both sales (~$725K) and profit ($108K)
- Central is the weakest region by profit margin (7.9%)
- All regions show consistent sales patterns

### 2. Category & Sub-Category
- Technology and Office Supplies are profitable across regions
- Furniture has high sales but low profit
- Tables (-$17,725), Bookcases (-$3,472), and Supplies (-$1,189) are loss-making

### 3. Discount Impact
- Orders with no discount: avg profit $66.90
- Orders with >20% discount: avg profit becomes negative
- Heavy discounts (>40%) generate significant losses (-$100K+ combined)

### 4. Customer Segments
- Consumer drives ~50% of sales but has lowest average order value ($223)
- Home Office has highest AOV ($241) — key segment for premium sales
- Corporate is a balanced middle segment

### 5. Time Trends
- Sales peak in Q4 (Sep–Dec) each year — holiday season
- 2015 shows ~29% YoY growth vs 2014

### 6. Correlation Insights
- Discount ↔ Profit correlation: -0.219 (moderate negative)
- Sales ↔ Profit correlation: +0.479 (moderate positive)
